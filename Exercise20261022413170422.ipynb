````python

# 學生分數分析系統（Student Score Analyzer）

## 一、學習目標（Objectives）

透過本題，練習 Python 中 `*`、`**` 與 `return` 的多種用途，包括：

1. 使用 `*args` 收集不定數量的位置參數。
2. 使用 `**kwargs` 收集不定數量的關鍵字參數。
3. 使用 `*` 解包可迭代物件。
4. 使用 `**` 解包字典。
5. 理解 `*` 的乘法用途及 `**` 的次方用途。
6. 使用 `return` 回傳函數的計算結果。

## 二、題目說明（Problem Description）

請設計一個 Python 程式，建立學生分數分析系統。

程式必須能夠接收多筆分數，根據指定的倍率（multiplier）及加分值（bonus）調整分數，接著計算調整後的分數平方、總分及平均分數，最後回傳分析結果。

### Part 1：完成函數

請閱讀以下函數，理解各部分的功能。

```python
def analyze_scores(*scores, **options):
    """
    分析學生分數。

    Parameters:
        *scores: 接收不定數量的分數
        **options: 接收選擇性設定，例如 multiplier 和 bonus

    Returns:
        一個字典，包含調整後的分數、平方、總分及平均分數。
    """
    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    adjusted = [
        score * multiplier + bonus
        for score in scores
    ]

    squared = [
        score ** 2
        for score in adjusted
    ]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average
    }
```

### Part 2：呼叫函數

已知下列資料：

```python
scores = [60, 70, 80]

settings = {
    "multiplier": 1.1,
    "bonus": 5
}
```

**問題 1：** 請使用 `*` 與 `**` 解包 `scores` 和 `settings`，呼叫 `analyze_scores()` 函數，並將結果存入變數 `result`。

```python
result = analyze_scores(*scores, **settings)
```



**問題 2：** 請印出 `result`，觀察函數回傳的結果。

```python
print(result)
```



### Part 3：思考與分析（Questions）

1. 在函數內部，`scores` 的資料型態是什麼？元組 (tuple)其值為何？ (60, 70, 80)

2. 在函數內部，`options` 的資料型態是什麼？字典 (dict)其值為何？{'multiplier': 1.1, 'bonus': 5}

3. `score * multiplier` 與 `score ** 2` 有什麼不同？請說明兩個運算子的意義。
	 * 是 乘法運算子：score * multiplier 代表將分數乘以指定的倍率。
	 ** 是 次方運算子：score ** 2 代表計算分數的二次方（平方）。


4. 為什麼函數使用 `return`，而不是只使用 `print()`？• print() 只是將結果顯示在螢幕上，無法讓程式的其他部分重複利用該數據。
 return 會將計算結果回傳給呼叫端，讓結果可以被儲存在變數中（如 result）、傳遞給其他函數，或進行進一步的分析與存檔

5. 如果呼叫 `analyze_scores(60, 70, 80)`，但沒有提供 `settings`，程式會如何執行？請說明原因。
	程式仍會正常執行，不會報錯。
	原因： 函數內部使用了字典的 .get() 方法，並設定了預設值：options.get("multiplier", 1) 和 options.get("bonus", 0)。當未提供 settings 時，options 為空字典，系統會自動採用預設值（乘以 1、加 0），因此調整後的分數會與原分數相同。


6. 請修改函數，讓回傳的字典額外包含調整後的最高分數，鍵值名稱為 `"max_score"`。
def analyze_scores(*scores, **options):
    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    adjusted = [score * multiplier + bonus for score in scores]
    squared = [score ** 2 for score in adjusted]
    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    max_score = max(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average,
        "max_score": max_score  
    }


7. 如果沒有傳入任何分數，程式會回傳什麼結果？請解釋程式如何處理這種情況。 
程式會回傳所有列表均為空、總分與平均皆為 0 的字典（例如：{'adjusted_scores': [], 'squared_scores': [], 'total': 0, 'average': 0}）。
 處理： 程式使用了條件式運算（三元運算子） total / len(adjusted) if adjusted else 0。當沒有傳入分數時，adjusted 為空列表（在布林判斷中為 False），程式會直接將 average 設為 0，從而避免了除以零（ZeroDivisionError）的程式崩潰危機。


### Part 4：挑戰題（Challenge）

請修改函數，使其也能透過名為 `scores_list` 的關鍵字參數接收一個分數串列。

例如：

```python
result = analyze_scores(
    multiplier=1.1,
    bonus=5,
    scores_list=[60, 70, 80]
)
```

請思考並回答：

1. 原本的 `*scores` 與 `scores_list` 有什麼不同？ *scores 是位置參數的收集器，它期待接收「多個獨立的參數」（如 60, 70, 80），並自動將它們打包成一個 tuple。
scores_list 是一個關鍵字參數，它期待接收「單一個包含多個分數的迭代物件（如串列 [60, 70, 80]）」。
2. 為什麼原本的函數無法直接使用上述呼叫方式處理 `scores_list`？ 
	因為呼叫時 scores_list=[60, 70, 80] 是以關鍵字參數形式傳入，它會被打包進 **options 字典中（變成 options['scores_list']），此時 *scores 接收不到任何資料（變成空 tuple），導致運算邏輯無法找到分數。

3. 請修改函數，使上述呼叫方式也能正確計算結果。
def analyze_scores(*scores, **options):
    # 優先從關鍵字參數 options 中尋找 scores_list，若沒有則使用位置參數 scores
    actual_scores = options.get("scores_list", scores)

    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    adjusted = [score * multiplier + bonus for score in actual_scores]
    squared = [score ** 2 for score in adjusted]
    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average
    }


## 三、語法整理（Learning Focus）

| 語法 | 用途 |
|---|---|
| `*scores`（函數定義） | 將不定數量的位置參數收集成 tuple |
| `**options`（函數定義） | 將不定數量的關鍵字參數收集成 dictionary |
| `score * multiplier` | 乘法 |
| `score ** 2` | 次方 |
| `f(*scores)` | 將可迭代物件解包為位置參數 |
| `f(**settings)` | 將字典解包為關鍵字參數 |
| `return` | 將函數的結果回傳給呼叫端 |

## 四、總結

請根據本題，說明 `*`、`**` 與 `return` 在 Python 中的不同用途，並舉出至少一個程式碼例子，展示同一個符號在不同語法位置所代表的意義。
# 範例：展示 * 與 ** 的多元身份

# 1. 在函數【定義】中：* 和 ** 用於【打包/收集】參數
def make_sandwich(*ingredients, **sauces):
    print(ingredients)  # 變成 tuple
    print(sauces)       # 變成 dict

# 2. 在數學運算中：* 是乘法，** 是次方
result = 5 * 2 ** 3     # 先計算 2 的 3 次方(8)，再乘以 5，結果為 40

# 3. 在函數【呼叫】中：* 和 ** 用於【解包】資料
food_list = ["火腿", "起司"]
sauce_dict = {"mayo": "多一點", "mustard": "少一點"}

# 這裡的 * 和 ** 會把串列與字典拆開來丟進函數中
make_sandwich(*food_list, **sauce_dict)
````
