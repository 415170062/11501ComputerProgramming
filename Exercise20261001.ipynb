def get_zodiac(year):
    # 判斷是否為民國年份
    # 民國 1 年等於西元 1912 年
    if year <= 1911:
        # 假設小於等於 1911 的輸入為民國年份（民國元年前為負數或不存在，此處視為民國）
        # 如果使用者輸入 100，代表民國 100 年，換算成西元為 100 + 1911 = 2011 年
        ad_year = year + 1911
        year_type = f"民國 {year} 年"
    else:
        # 大於 1911 視為西元年份
        ad_year = year
        year_type = f"西元 {year} 年"
    
    # 十二生肖對應表（以西元年份除以 12 的餘數來決定）
    # 西元 0 年（或 12 年）基準點為「猴」
    zodiacs = {
        0: "猴 (monkey)",
        1: "雞 (rooster)",
        2: "狗 (dog)",
        3: "豬 (pig)",
        4: "鼠 (rat)",
        5: "牛 (ox)",
        6: "虎 (tiger)",
        7: "兔 (rabbit)",
        8: "龍 (dragon)",
        9: "蛇 (snake)",
        10: "馬 (horse)",
        11: "羊 (sheep)"
    }
    
    remainder = ad_year % 12
    return year_type, zodiacs[remainder]

# 執行主程式
try:
    user_input = int(input("請輸入年份（西元或民國，例如 2026 或 115）："))
    if user_input <= 0:
        print("請輸入大於 0 的有效年份。")
    else:
        y_type, zodiac = get_zodiac(user_input)
        print(f"您輸入的是【{y_type}】，對應的生肖為：{zodiac}")
except ValueError:
