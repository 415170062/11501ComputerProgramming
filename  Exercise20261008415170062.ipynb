```python
 使用者輸入：請輸入正整數$n$，其中$1\le n\le 9$，輸出$nn$乘法表。格式如Tranclass的圖片Exercise20261008.png，以$n=5$為例。

1. 只能使用for。
3. 字串格式化只能使用 f-string 對齊。
```

```python
n = int(input("請輸入正整數n (1 <= n <= 9): "))

for i in range(1, n + 1):
    for j in range(1, n + 1):
        
        space = " " if j < n else ""
        
        print(f"{i}×{j}={i*j}{space}", end="")
    print()
```

```python
使用者輸入：請輸入正整數$n$，其中$1\le n\le 9$，輸出二進位$nn$乘法表。格式如如Tranclass的圖片Exercise20261008.png，以$n=5$為例。                                                                                                         
                                                                                                            
1. 只能使用while。                                                                                            
3. 字串格式化只能使用 f-string 對齊。                                                                       
```

```python
n = int(input("請輸入正整數n (1 <= n <= 9): "))

i = 1
while i <= n:
    j = 1
    while j <= n:
       
        space = " " if j < n else ""
        
        print(f"{i:03b}×{j:03b}={i*j:06b}{space}", end="")
        j += 1
    print()  # 換行
    i += 1
```
